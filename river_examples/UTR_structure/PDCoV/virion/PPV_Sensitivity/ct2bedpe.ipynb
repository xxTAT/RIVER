```python
ct_file = 'RN7SL1_99_v2.ct'
chrom = 'NR_002715.1'
ct = open(ct_file, "r")

# 基于CT文件名生成基础BEDPE文件名
base_bedpe_name = ct_file.split('.')[0]  # 去掉扩展名
res = None
start = 0

# 初始化变量
current_structure = None
out = None

for index, row in enumerate(ct):
    row = row.strip()
    if not row:  # 跳过空行
        continue
    if 'ENERGY' in row:  # 假设结构之间以包含"ENERGY"的行分隔
        if out is not None:
            out.close()
        # 为每个新结构生成一个新的BEDPE文件
        structure_index = 1 if current_structure is None else current_structure + 1
        bedpe = f'{base_bedpe_name}_structure{structure_index}.bedpe'
        out = open(bedpe, "w")
        out.write('chrom1\tstart1\tend1\tchrom2\tstart2\tend2\n')
        current_structure = structure_index
        continue
    #if index == 0:
    #    continue  # 跳过CT文件的标题行
    sub = row.split()
    if len(sub) < 6:
        continue
    if int(sub[4]) != 0:
        if not res:
            out.write(chrom+"\t" + str(int(sub[0]) - 1 + start) + "\t" + str(int(sub[0]) + start) + "\t"+chrom+"\t" + str(int(sub[4]) - 1 + start) + "\t" + str(int(sub[4]) + start) + "\n")
        else:
            out.write(chrom+"\t" + str(((int(sub[0]) - 1) // res) * res) + "\t" + str(((int(sub[0]) - 1) // res) * res + res) + "\t"+chrom+"\t" + str(((int(sub[4]) - 1) // res) * res) + "\t" + str(((int(sub[4]) - 1) // res) * res + res) + "\n")

# 关闭最后一个文件
if out is not None:
    out.close()
ct.close()
```
